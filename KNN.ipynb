# KNN\nOriginal educational implementation using scikit-learn datasets.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
scores = {}
for k in range(1, 16, 2):
    model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=k))
    scores[k] = cross_val_score(model, X_train, y_train, cv=5).mean()
best_k = max(scores, key=scores.get)
model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=best_k))
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print('Selected k:', best_k)
print(classification_report(y_test, predictions, target_names=load_iris().target_names))
ConfusionMatrixDisplay.from_predictions(y_test, predictions)
plt.show()
```
